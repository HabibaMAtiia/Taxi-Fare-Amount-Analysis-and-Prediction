```python
# Find the top 10 users based on the total JFK distance
top_10_users_jfk = data.groupby('User Name')['jfk_dist'].sum().nlargest(10).index

# Filter the data to include only the top 10 users
top_10_data_jfk = data[data['User Name'].isin(top_10_users_jfk)]

# Bar Plot of Total JFK Distance for Top 10 Users
plt.figure(figsize=(14, 8))
total_jfk_distance_by_user = top_10_data_jfk.groupby('User Name')['jfk_dist'].sum().reset_index()
sns.barplot(x='jfk_dist', y='User Name', data=total_jfk_distance_by_user.sort_values(by='jfk_dist', ascending=False))
plt.title('Total JFK Distance for Top 10 Users')
plt.xlabel('Total JFK Distance')
plt.ylabel('User Name')
plt.show()
```

```python
# Filter out trips where jfk_dist is greater than 0
jfk_trips = data[data['jfk_dist'] > 0]

# Find the top 10 users based on the number of trips to/from JFK
top_10_users_jfk_trips = jfk_trips['User Name'].value_counts().nlargest(10).index

# Filter the data to include only the top 10 users
top_10_jfk_trip_data = jfk_trips[jfk_trips['User Name'].isin(top_10_users_jfk_trips)]

# Count the number of JFK trips for each of the top 10 users
jfk_trip_counts = top_10_jfk_trip_data['User Name'].value_counts().reset_index()
jfk_trip_counts.columns = ['User Name', 'JFK Trip Count']

# Bar Plot of Number of JFK Trips for Top 10 Users
plt.figure(figsize=(14, 8))
sns.barplot(x='JFK Trip Count', y='User Name', data=jfk_trip_counts.sort_values(by='JFK Trip Count', ascending=False))
plt.title('Number of Trips to/from JFK for Top 10 Users')
plt.xlabel('Number of JFK Trips')
plt.ylabel('User Name')
plt.show()
```

```python
# Find the top 10 drivers based on the number of trips to/from JFK
top_10_drivers_jfk_trips = jfk_trips['Driver Name'].value_counts().nlargest(10).index

# Filter the data to include only the top 10 drivers
top_10_jfk_trip_data = jfk_trips[jfk_trips['Driver Name'].isin(top_10_drivers_jfk_trips)]

# Count the number of JFK trips for each of the top 10 drivers
jfk_trip_counts = top_10_jfk_trip_data['Driver Name'].value_counts().reset_index()
jfk_trip_counts.columns = ['Driver Name', 'JFK Trip Count']

# Bar Plot of Number of JFK Trips for Top 10 Drivers
plt.figure(figsize=(14, 8))
sns.barplot(x='JFK Trip Count', y='Driver Name', data=jfk_trip_counts.sort_values(by='JFK Trip Count', ascending=False))
plt.title('Number of Trips to/from JFK for Top 10 Drivers')
plt.xlabel('Number of JFK Trips')
plt.ylabel('Driver Name')
plt.show()
```

```python
total_passenger_count_by_driver = data.groupby('Driver Name')['passenger_count'].sum().reset_index()

# Sort the data by 'passenger_count' in descending order and get the top 10 drivers
top_10_drivers_passenger_count = total_passenger_count_by_driver.sort_values(by='passenger_count', ascending=False).head(10)

# Bar Plot of Total Passenger Count for Top 10 Drivers
plt.figure(figsize=(14, 8))
sns.barplot(x='passenger_count', y='Driver Name', data=top_10_drivers_passenger_count)
plt.title('Total Passenger Count for Top 10 Drivers')
plt.xlabel('Total Passenger Count')
plt.ylabel('Driver Name')
plt.show()
```
